```python


#ler excel
caminho_pasta = r"C:\Users\rodri\Downloads\PedidosEmprestimo.xlsx"

#Carregar a planilha
ex = pd.read_excel(caminho_pasta)

#Mostrar os nomes das colunas

#print("Colunas disponíveis:", df.columns)

#Acessar clunas
celula1 = ex[ex.columns[0]].iloc[0]  
celula2 = ex[ex.columns[1]].iloc[0]  
celula3 = ex[ex.columns[2]].iloc[0]
celula4 = ex[ex.columns[3]].iloc[0]
celula5 = ex[ex.columns[4]].iloc[0]

#print("Linha:", celula1,celula2,celula3,celula4,celula5)

```

```python
#Acessar formulario web
Formulario = webdriver.Chrome()
Formulario.get("https://uibank.uipath.com/loans/apply")
```

```python
#Preenchendo forumulario 

email = Formulario.find_element(By.ID, "email")
email.clear()
email.send_keys(celula1)

time.sleep(1)


valor_emprestimo = Formulario.find_element(By.ID, "amount")
valor_emprestimo.clear()
valor_emprestimo.send_keys(str(celula2))

time.sleep(1)


prazo_emprestimo = Formulario.find_element(By.XPATH, '//*[@id="term"]').send_keys(str(celula3))

time.sleep(1)

renda_atual = Formulario.find_element(By.ID, "income")
renda_atual.clear()
renda_atual.send_keys(str(celula4))

time.sleep(1)

idade = Formulario.find_element(By.ID, "age")
idade.clear()
idade.send_keys(str(celula5))

time.sleep(1)

Formulario.find_element(By.ID, "submitButton").click()

```

```python
#Retorno da solicitaçao realizada
time.sleep(5)
#Achar os elemntos h1
titulos = Formulario.find_elements(By.TAG_NAME, "h1")  
text = [titulo.text for titulo in titulos]  
#exibir cada texto encontrado na lista h1
for texto in text:
    print(texto)  

Formulario.quit()
```
